### Project Overview
This beginner-level mini project uses **Pandas DataFrames** to store and manage book records and **NumPy arrays** to perform statistical analysis (average copies, fine calculations, and stock metrics).

```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Step 2: Initialize the Library Database using Pandas DataFrame
initial_data = {
    'Book_ID': [101, 102, 103, 104, 105],
    'Title': [
        'Python Programming',
        'Engineering Mathematics',
        'Data Structures with C',
        'Basic Electrical Engineering',
        'Physics for Engineers'
    ],
    'Author': ['Guido Van', 'B.S. Grewal', 'E. Balagurusamy', 'V.K. Mehta', 'H.K. Malik'],
    'Total_Copies': [10, 15, 8, 12, 7],
    'Issued_Copies': [3, 5, 2, 4, 1],
    'Price_INR': [450, 620, 500, 480, 550]
}

library_df = pd.DataFrame(initial_data)
library_df['Available_Copies'] = library_df['Total_Copies'] - library_df['Issued_Copies']

print("--- Initial Library Inventory ---")
display(library_df)
```

```python
# Step 3: Define Core Library Operations

def view_books(df):
    print("\n=== CURRENT LIBRARY BOOKS ===")
    display(df)

def add_book(df, book_id, title, author, total_copies, price):
    if book_id in df['Book_ID'].values:
        print(f"Error: Book ID {book_id} already exists!")
        return df
    new_row = pd.DataFrame([{
        'Book_ID': book_id,
        'Title': title,
        'Author': author,
        'Total_Copies': total_copies,
        'Issued_Copies': 0,
        'Price_INR': price,
        'Available_Copies': total_copies
    }])
    df = pd.concat([df, new_row], ignore_index=True)
    print(f"Book '{title}' added successfully!")
    return df

def search_book(df, keyword):
    result = df[df['Title'].str.contains(keyword, case=False, na=False) | 
                df['Author'].str.contains(keyword, case=False, na=False)]
    if result.empty:
        print(f"No books found matching '{keyword}'.")
    else:
        print(f"\n=== Search Results for '{keyword}' ===")
        display(result)

def issue_book(df, book_id):
    if book_id not in df['Book_ID'].values:
        print("Invalid Book ID!")
        return df
    idx = df.index[df['Book_ID'] == book_id][0]
    if df.loc[idx, 'Available_Copies'] > 0:
        df.loc[idx, 'Issued_Copies'] += 1
        df.loc[idx, 'Available_Copies'] -= 1
        print(f"Book '{df.loc[idx, 'Title']}' issued successfully!")
    else:
        print("Sorry, no copies available to issue.")
    return df

def return_book(df, book_id, days_late=0):
    if book_id not in df['Book_ID'].values:
        print("Invalid Book ID!")
        return df
    idx = df.index[df['Book_ID'] == book_id][0]
    if df.loc[idx, 'Issued_Copies'] > 0:
        df.loc[idx, 'Issued_Copies'] -= 1
        df.loc[idx, 'Available_Copies'] += 1
        # Fine calculation using NumPy
        fine_per_day = 5  # Rs. 5 per day
        fine = np.maximum(0, days_late * fine_per_day)
        print(f"Book '{df.loc[idx, 'Title']}' returned! Late Fine: Rs. {fine}")
    else:
        print("Error: No issued copies of this book to return.")
    return df
```

```python
# Step 4: Library Statistical Analysis using NumPy
def library_analytics(df):
    total_books_arr = df['Total_Copies'].to_numpy()
    issued_books_arr = df['Issued_Copies'].to_numpy()
    prices_arr = df['Price_INR'].to_numpy()
    
    total_stock = np.sum(total_books_arr)
    total_issued = np.sum(issued_books_arr)
    avg_price = np.mean(prices_arr)
    total_inventory_value = np.sum(total_books_arr * prices_arr)
    utilization_rate = (total_issued / total_stock) * 100
    
    print("\n=== LIBRARY ANALYTICS (POWERED BY NUMPY) ===")
    print(f"Total Books in Library      : {total_stock}")
    print(f"Total Books Currently Issued: {total_issued}")
    print(f"Average Book Price          : Rs. {avg_price:.2f}")
    print(f"Total Inventory Value       : Rs. {total_inventory_value}")
    print(f"Library Utilization Rate    : {utilization_rate:.2f}%")
```

```python
# Step 5: Demonstration of All Features (Automated Test Run)
library_df = add_book(library_df, 106, 'Introduction to NumPy', 'Travis Oliphant', 9, 400)
search_book(library_df, 'Python')
library_df = issue_book(library_df, 101)
library_df = return_book(library_df, 102, days_late=3)
view_books(library_df)
library_analytics(library_df)
```

```python
# Step 6: Interactive Menu-Driven System (Run this cell to use interactively)
def run_library_system(df):
    while True:
        print("\n=========================================")
        print("      LIBRARY MANAGEMENT SYSTEM MENU     ")
        print("=========================================")
        print("1. View All Books")
        print("2. Add a New Book")
        print("3. Search Book by Title/Author")
        print("4. Issue a Book")
        print("5. Return a Book")
        print("6. View NumPy Library Statistics")
        print("7. Save Data to CSV & Exit")
        
        choice = input("Enter your choice (1-7): ")
        
        if choice == '1':
            view_books(df)
        elif choice == '2':
            try:
                b_id = int(input("Enter Book ID: "))
                title = input("Enter Book Title: ")
                author = input("Enter Author Name: ")
                copies = int(input("Enter Total Copies: "))
                price = float(input("Enter Price (INR): "))
                df = add_book(df, b_id, title, author, copies, price)
            except ValueError:
                print("Invalid input! Please enter numeric values for ID, Copies, and Price.")
        elif choice == '3':
            kw = input("Enter Title or Author keyword to search: ")
            search_book(df, kw)
        elif choice == '4':
            try:
                b_id = int(input("Enter Book ID to issue: "))
                df = issue_book(df, b_id)
            except ValueError:
                print("Please enter a valid numeric Book ID.")
        elif choice == '5':
            try:
                b_id = int(input("Enter Book ID to return: "))
                late = int(input("Enter days late (0 if on time): "))
                df = return_book(df, b_id, late)
            except ValueError:
                print("Please enter valid numeric inputs.")
        elif choice == '6':
            library_analytics(df)
        elif choice == '7':
            df.to_csv('library_records.csv', index=False)
            print("Records saved to 'library_records.csv'. Exiting system. Thank you!")
            break
        else:
            print("Invalid choice! Please select between 1 and 7.")
    return df

# Uncomment the line below to run the interactive menu in Google Colab:
# library_df = run_library_system(library_df)
```
